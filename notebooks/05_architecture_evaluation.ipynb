```python
# Cell 6

def normalize_value(value):
    """
    Normalize extracted values before comparison.
    """

    if value is None:
        return ""

    value = str(value)

    value = value.strip().lower()

    value = " ".join(
        value.split()
    )

    return value


print("Normalization function created successfully.")
```

```python
# Cell 7

def compare_extraction(
    extracted,
    expected
):
    """
    Compare extracted values against
    expected ground-truth values.
    """

    extracted_set = {
        normalize_value(value)
        for value in extracted
    }

    expected_set = {
        normalize_value(value)
        for value in expected
    }

    true_positives = (
        extracted_set & expected_set
    )

    false_positives = (
        extracted_set - expected_set
    )

    false_negatives = (
        expected_set - extracted_set
    )

    return {
        "true_positives": true_positives,
        "false_positives": false_positives,
        "false_negatives": false_negatives
    }


print("Extraction comparison function created successfully.")
```

```python
# Cell 8

def calculate_metrics(
    comparison
):
    """
    Calculate precision, recall and F1
    from extraction comparison results.
    """

    tp = len(
        comparison["true_positives"]
    )

    fp = len(
        comparison["false_positives"]
    )

    fn = len(
        comparison["false_negatives"]
    )

    precision = (
        tp / (tp + fp)
        if (tp + fp) > 0
        else 0.0
    )

    recall = (
        tp / (tp + fn)
        if (tp + fn) > 0
        else 0.0
    )

    if precision + recall > 0:

        f1 = (
            2 * precision * recall
            / (precision + recall)
        )

    else:

        f1 = 0.0

    return {
        "true_positives": tp,
        "false_positives": fp,
        "false_negatives": fn,
        "precision": precision,
        "recall": recall,
        "f1": f1
    }


print("Metric calculation function created successfully.")
```

```python
# Cell 9

def extract_entity_names(items):
    """
    Extract comparable names from structured
    architecture records.
    """

    names = []

    for item in items:

        if isinstance(item, dict):

            name = item.get("name")

            if name:
                names.append(name)

        else:

            names.append(item)

    return names


component_extracted_names = extract_entity_names(
    architecture_knowledge["components"]
)

component_comparison = compare_extraction(
    component_extracted_names,
    ground_truth["components"]
)

component_metrics = calculate_metrics(
    component_comparison
)

print("=" * 70)
print("COMPONENT EXTRACTION EVALUATION")
print("=" * 70)

print(
    "\nExtracted component names:"
)

for name in component_extracted_names:
    print("  -", name)

print(
    "\nTrue Positives:",
    component_metrics["true_positives"]
)

print(
    "False Positives:",
    component_metrics["false_positives"]
)

print(
    "False Negatives:",
    component_metrics["false_negatives"]
)

print(
    f"\nPrecision: "
    f"{component_metrics['precision']:.2%}"
)

print(
    f"Recall: "
    f"{component_metrics['recall']:.2%}"
)

print(
    f"F1 Score: "
    f"{component_metrics['f1']:.2%}"
)
```

```python
# Cell 18

import re


def normalize_functional_flow(value):
    """
    Normalize functional-flow descriptions
    while preserving their meaning.
    """

    value = normalize_value(value)

    # Remove punctuation
    value = re.sub(r"[^\w\s]", "", value)

    # Remove harmless article differences
    value = re.sub(
        r"\bthe\b",
        "",
        value
    )

    # Normalize whitespace again
    value = " ".join(value.split())

    return value


def compare_functional_flows(
    extracted,
    expected
):

    extracted_set = {
        normalize_functional_flow(value)
        for value in extracted
    }

    expected_set = {
        normalize_functional_flow(value)
        for value in expected
    }

    true_positives = (
        extracted_set & expected_set
    )

    false_positives = (
        extracted_set - expected_set
    )

    false_negatives = (
        expected_set - extracted_set
    )

    return {
        "true_positives": true_positives,
        "false_positives": false_positives,
        "false_negatives": false_negatives
    }


functional_flow_extracted = (
    extract_functional_flow_descriptions(
        architecture_knowledge[
            "functional_flows"
        ]
    )
)

functional_flow_comparison = (
    compare_functional_flows(
        functional_flow_extracted,
        ground_truth[
            "functional_flows"
        ]
    )
)

functional_flow_metrics = calculate_metrics(
    functional_flow_comparison
)


print("=" * 70)
print("FUNCTIONAL FLOW EXTRACTION EVALUATION")
print("=" * 70)

print("\nExtracted functional flows:")

for flow in functional_flow_extracted:
    print("  -", flow)

print(
    "\nTrue Positives:",
    functional_flow_metrics[
        "true_positives"
    ]
)

print(
    "False Positives:",
    functional_flow_metrics[
        "false_positives"
    ]
)

print(
    "False Negatives:",
    functional_flow_metrics[
        "false_negatives"
    ]
)

print(
    f"\nPrecision: "
    f"{functional_flow_metrics['precision']:.2%}"
)

print(
    f"Recall: "
    f"{functional_flow_metrics['recall']:.2%}"
)

print(
    f"F1 Score: "
    f"{functional_flow_metrics['f1']:.2%}"
)
```

```python
# Cell 19

def extract_architecture_relationships(items):
    """
    Extract source -> target architecture
    relationships.
    """

    relationships = []

    for item in items:

        if isinstance(item, dict):

            source = item.get(
                "source_component"
            )

            target = item.get(
                "target_component"
            )

            if source and target:

                relationships.append(
                    f"{source} -> {target}"
                )

        else:

            relationships.append(item)

    return relationships


relationship_extracted = (
    extract_architecture_relationships(
        architecture_knowledge[
            "architecture_relationships"
        ]
    )
)

relationship_comparison = compare_extraction(
    relationship_extracted,
    ground_truth[
        "architecture_relationships"
    ]
)

relationship_metrics = calculate_metrics(
    relationship_comparison
)

print("=" * 70)
print("ARCHITECTURE RELATIONSHIP EVALUATION")
print("=" * 70)

print("\nExtracted relationships:")

for relationship in relationship_extracted:
    print("  -", relationship)

print(
    "\nTrue Positives:",
    relationship_metrics[
        "true_positives"
    ]
)

print(
    "False Positives:",
    relationship_metrics[
        "false_positives"
    ]
)

print(
    "False Negatives:",
    relationship_metrics[
        "false_negatives"
    ]
)

print(
    f"\nPrecision: "
    f"{relationship_metrics['precision']:.2%}"
)

print(
    f"Recall: "
    f"{relationship_metrics['recall']:.2%}"
)

print(
    f"F1 Score: "
    f"{relationship_metrics['f1']:.2%}"
)
```

```python
# Cell 26

def check_architecture_consistency(
    architecture
):

    issues = []

    component_names = {
        item.get("name")
        for item in architecture.get(
            "components",
            []
        )
        if isinstance(item, dict)
    }

    referenced_entities = set()

    # Check dependencies
    for dependency in architecture.get(
        "dependencies",
        []
    ):

        if not isinstance(
            dependency,
            dict
        ):
            continue

        dependent = dependency.get(
            "dependent_component"
        )

        dependency_component = dependency.get(
            "dependency_component"
        )

        if dependent:
            referenced_entities.add(
                dependent
            )

        if dependency_component:
            referenced_entities.add(
                dependency_component
            )

    # Check architecture relationships
    for relationship in architecture.get(
        "architecture_relationships",
        []
    ):

        if not isinstance(
            relationship,
            dict
        ):
            continue

        source = relationship.get(
            "source_component"
        )

        target = relationship.get(
            "target_component"
        )

        if source:
            referenced_entities.add(
                source
            )

        if target:
            referenced_entities.add(
                target
            )

    # Identify referenced components that are
    # not part of the internal component inventory.
    external_entities = (
        referenced_entities
        - component_names
    )

    for entity in sorted(
        external_entities
    ):

        issues.append({
            "type": "external_component",
            "entity": entity,
            "context": "architecture"
        })

    return issues, component_names


consistency_issues, component_names = (
    check_architecture_consistency(
        architecture_knowledge
    )
)

print("=" * 70)
print("ARCHITECTURE CONSISTENCY CHECK")
print("=" * 70)

print(
    "\nInternal components:",
    sorted(component_names)
)

print(
    "\nIssues found:",
    len(consistency_issues)
)

for issue in consistency_issues:

    print(
        f"\n- {issue['type']}"
    )

    print(
        f"  Entity: {issue['entity']}"
    )

    print(
        f"  Context: {issue['context']}"
    )
```
